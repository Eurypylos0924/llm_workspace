whisper-1 모델이란?  
whisper-1은 오디오 파일에 포함된 음성을 텍스트로 변환하는 모델이다.  
- STT: 음성 데이터를 텍스트로 변환한다.
- 다국어 지원: 한국어, 영어 등 다양한 언어를 인식한다.
- 자동 언어 감지: 언어를 지정하지 않아도 음성의 언어를 자동으로 판별할 수 있다.
- 번역: 음성을 영어 텍스트로 번역하는 기능도 지원한다.


|매개변수	|설명|
|---       |---|
|model="whisper-1"      	|사용할 STT 모델 지정               |
|file=audio_file  	        |변환할 오디오 파일                 |
|language="ko"	            |음성 언어를 한국어로 지정           |
|response_format="text"	    |인식 결과를 일반 텍스트로 반환       |
|temperature=0	             |생성 과정의 무작위성을 낮추도록 설정 |

유튜브 영상 내려받기

````python
import yt_dlp
import os
#Python 표준 라이브러리로, 파일 복사, 이동, 삭제, 디렉터리 관리 및 실행 파일 경로 검색
import shutil

def download_youtube_mp3(video_id: str = 'vHdKMkIH7qM'):
    # FFmpeg 설치 여부 확인
    if shutil.which("ffmpeg") is None:
        print("오류: FFmpeg이 설치되어 있지 않습니다. 설치 후 다시 시도하세요.")
        return None

    url = f"https://www.youtube.com/watch?v={video_id}"
    output_dir = "downloads"
    os.makedirs(output_dir, exist_ok=True)  # 다운로드 폴더 생성

    # 동영상의 오디오를 다운로드하고 MP3 파일로 변환하기 위한 옵션을 정의한 딕셔너리
    ydl_opts = {
         # 다운로드할 미디어 형식을 지정한다.
         # bestaudio: 가장 좋은 품질의 오디오 스트림을 선택한다.
         # best: 오디오 전용 스트림이 없으면 최상의 통합 스트림을 선택한다.
        'format': 'bestaudio/best',
         # 다운로드 파일의 저장 경로와 이름을 지정한다.
         # output_dir: 파일을 저장할 디렉터리
         # %(id)s: 동영상 고유 ID
         # %(ext)s: 다운로드한 원본 파일의 확장자
        'outtmpl': f"{output_dir}/%(id)s.%(ext)s",
        # 다운로드가 완료된 후 수행할 후처리 작업을 지정한다.
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',    # FFmpeg을 사용하여 오디오를 추출하고 변환한다.
            'preferredcodec': 'mp3',        # 최종 오디오 파일을 MP3 형식으로 변환한다.
            'preferredquality': '192',     # MP3 인코딩 품질을 192kbps로 지정한다.
        }],
    }

# ydl_opts 에 따라서 음성만 뽑아낸다.
    try:
        # YoutubeDL 객체를 생성한다.
        # ydl_opts에는 다운로드 형식, 저장 경로,
        # MP3 변환 등의 설정이 포함되어 있다.
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
           # 지정한 URL의 동영상 정보를 가져온다.
           # download=True이므로 실제 다운로드도 수행한다.
           # 설정된 postprocessors에 따라 MP3 변환까지 수행한다.
           # info에는 동영상 ID, 제목, 길이 등의 메타데이터가 포함된다.
            info = ydl.extract_info(url, download=True)

            # 동영상 고유 ID를 이용하여 MP3 파일명을 생성한다.           
            file_name = f"{info['id']}.mp3"
            # 저장 디렉터리와 파일명을 결합하여 전체 경로를 생성한다.
            file_path = os.path.join(output_dir, file_name)

         # 다운로드 및 변환이 정상적으로 완료되면 경로를 출력한다.
        print(f"다운로드 완료! 파일 경로: {file_path}")

        # 생성된 MP3 파일의 경로를 반환한다.
        # 반환된 경로는 STT 모델의 입력 등에 사용할 수 있다.
        return file_path
    # 다운로드, 네트워크, FFmpeg 변환 등의 오류를 처리한다.
    except Exception as e:
        print(f"오류 발생: {e}")
        return None

#https://www.youtube.com/watch?v=vHdKMkIH7qM&list=RDvHdKMkIH7qM&start_radio=1
# 유튜브 아이디 : vHdKMkIH7qM
# 테스트 실행 vHdKMkIH7qM
audio_path = download_youtube_mp3('vHdKMkIH7qM')
print(audio_path)
````

파일 자르기

````python
import subprocess
import os

def trim_audio_file(file_path:str, max_size:int=24*1024*1024):
    '''오디오 파일을 24MB 크기로 자릅니다.'''
    # 앞 단계(다운로드)가 실패하면 file_path 가 None 이 된다.
    if not file_path:
        raise ValueError("오디오 파일 경로가 없습니다. 위의 다운로드 셀 출력의 오류 메시지를 확인하세요.")
    file_size = os.path.getsize(file_path)  # 바이트 단위
    output_path = file_path
    if file_size >= max_size:
        # ffprobe 로 파일의 비트레이트(bps)를 조회한다.
        bitrate = int(subprocess.run(
            ["ffprobe", "-v", "error", "-show_entries", "format=bit_rate",
             "-of", "default=noprint_wrappers=1:nokey=1", file_path],
            capture_output=True, text=True, check=True,
        ).stdout.strip())

        # 최대 24MB로 자를 수 있는 재생 시간 계산 (초)
        max_duration = max_size * 8 / bitrate

        # 출력 파일 경로 설정
        output_path = file_path.replace(".mp3", "_trimmed.mp3")

        # ffmpeg로 앞에서부터 max_duration 초만 복사해서 자르기 (재인코딩 없음)
        subprocess.run(
            ["ffmpeg", "-y", "-i", file_path, "-t", str(max_duration), "-c", "copy", output_path],
            capture_output=True, check=True,
        )
    return output_path

trimmed_audio_path = trim_audio_file(audio_path)
````

오디오 파일에서 텍스트 추출하기
- 파일을 서버에 보내서 텍스트를 추출하므로 실행시간이 좀 걸린다.
- OpenAI의 음성 인식(STT, Speech-to-Text) 모델인 Whisper

````python
def transcribe_audio(audio_path):
    audio_file= open(audio_path, "rb")
    response = client.audio.transcriptions.create(
        model="whisper-1",
        file=audio_file,
        response_format="text"
    )
    return response

# 텍스트 변환 실행
transcribed_text = transcribe_audio(trimmed_audio_path)
print("변환된 텍스트:\n", transcribed_text)
````

텍스트 내용 요약하기(gpt)

````python
def summarize_text(text):
  response = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0,
    messages=[
      { "role": "system", "content": "주어진 텍스트를 간결하고 명확하게 요약해 주세요." },
      { "role": "user", "content": text }
    ],
  )
  return response.choices[0].message.content

# 요약 실행
summary = summarize_text(transcribed_text)
print("요약된 내용:\n", summary)
````
