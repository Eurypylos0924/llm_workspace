````python
# (2) 브라우저 마이크로 N초 녹음 → /content/rec.webm 저장

import base64

def record_to_webm(seconds=5, path="/content/rec.webm"):
    js = f"""
    (async () => {{
      const sleep = ms => new Promise(r => setTimeout(r, ms));
      const b2text = blob => new Promise((resolve, reject) => {{
        const reader = new FileReader();
        reader.onloadend = () => resolve(reader.result);
        reader.onerror = reject;
        reader.readAsDataURL(blob);
      }});
      const stream = await navigator.mediaDevices.getUserMedia({{ audio: true }});
      const rec = new MediaRecorder(stream, {{ mimeType: 'audio/webm' }});
      const chunks = [];
      rec.ondataavailable = e => chunks.push(e.data);
      rec.start();
      await sleep({int(seconds*1000)});
      rec.stop();
      await new Promise(res => rec.onstop = res);
      stream.getTracks().forEach(t => t.stop());
      const blob = new Blob(chunks, {{ type: 'audio/webm' }});
      return await b2text(blob);   // data: URL (base64) 반환
    }})()
    """
    b64 = output.eval_js(js)
    data = base64.b64decode(b64.split(",")[1])
    with open(path, "wb") as f:
        f.write(data)
    return path

webm_path = record_to_webm(seconds=5, path="output/rec.webm")
print("Saved:", webm_path)
````

````python
from openai import OpenAI
from IPython.display import Audio, display
import base64, os, subprocess

# 1) OpenAI API 초기화
OPENAI_API_KEY = userdata.get('OPENAI_API_KEY')
assert OPENAI_API_KEY, "OPENAI_API_KEY가 설정되어 있어야 합니다."
client = OpenAI(api_key=OPENAI_API_KEY)

# 2) 브라우저 마이크로 N초 녹음 → /content/rec.webm 저장
def record_to_webm(seconds=5, path="/content/rec.webm"):
 
    b64 = output.eval_js(js)
    data = base64.b64decode(b64.split(",")[1])
    with open(path, "wb") as f:
        f.write(data)
    return path

# 3) webm → 16kHz mono WAV 변환 (Whisper 친화 형식)
def webm_to_wav(webm_path="/content/rec.webm", wav_path="/content/rec.wav"):
    subprocess.run(
        ["ffmpeg", "-y", "-i", webm_path, "-ar", "16000", "-ac", "1", wav_path],
        check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE
    )
    return wav_path

# 4) STT (Whisper) — 한국어 고정
def transcribe_audio(file_path):
    with open(file_path, "rb") as fp:
        tr = client.audio.transcriptions.create(
            model="whisper-1",
            file=fp,
            language="ko"      # ✅ 한국어 강제
            # temperature=0.0, # (선택) 더 보수적으로 인식
        )
    return tr.text.strip()

# 5) GPT 질의 — 한국어 응답 강제
def ask_gpt(prompt_text):
    resp = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"항상 한국어(존댓말)로 자연스럽고 간결하게 답변하세요. 필요한 경우만 1~2문장 추가 설명을 덧붙이세요."},
            {"role":"user","content": prompt_text}
        ]
    )
    return resp.choices[0].message.content.strip()

# 6) TTS (mp3 생성) + 자동 재생
def speak_text(text, output_file="/content/out.mp3"):
    speech = client.audio.speech.create(
        model="tts-1",
        voice="nova",
        input=text
    )
    with open(output_file, "wb") as f:
        f.write(speech.content)
    print("음성 파일 생성:", output_file)
    display(Audio(output_file, autoplay=True))

# 7) 전체 파이프라인
def main(duration_sec=5):
    print("▶ 브라우저 마이크 녹음 시작… (권한 허용 필요)")
    webm = record_to_webm(seconds=duration_sec, path="/content/rec.webm")
    wav = webm_to_wav(webm, "/content/rec.wav")
    print("녹음/변환 완료:", wav)

    question = transcribe_audio(wav)
    print("[STT] 질문 내용:", question)

    answer = ask_gpt(question)
    print("[GPT] 답변 내용:", answer)

    speak_text(answer, "/content/out.mp3")

# 실행
main(5)
````

````python
import os
import time
from openai import OpenAI
import pygame
from scipy.io.wavfile import write
import sounddevice as sd

client = OpenAI()

OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)


# 1. 안전한 MP3 재생 함수 (파일 잠금 완벽 해제)
def playmp3(output_file):
    print("🔊 답변 음성 재생 중...")
    pygame.mixer.init()
    pygame.mixer.music.load(output_file)
    pygame.mixer.music.play()

    while pygame.mixer.music.get_busy():
        time.sleep(0.1)

    # [핵심] 재생 완료 후 트랙 언로드 및 믹서 종료 (파일 잠금 해제)
    pygame.mixer.music.unload()
    pygame.mixer.quit()


# 2. TTS 변환 및 저장 함수 (타임스탬프 활용으로 파일 충돌 방지)
def speak_text(text):
    print("🗣️ 텍스트를 음성으로 변환 중 (TTS)...")

    # 충돌 방지를 위해 실행 시점의 타임스탬프로 파일명 생성
    filename = f"{OUTPUT_DIR}/out_{int(time.time())}.mp3"

    response = client.audio.speech.create(
        model="tts-1", voice="nova", input=text
    )

    # 최신 SDK 방식 파일 쓰기
    response.write_to_file(filename)
    print("✅ 음성 파일 생성 완료:", filename)

    # 재생
    playmp3(filename)


# 3. 음성 인식(STT) 예외 처리 보완
def transcribe_audio(file_path):
    print("🔍 음성을 텍스트로 변환하는 중 (STT)...")
    with open(file_path, "rb") as fp:
        transcription = client.audio.transcriptions.create(
            file=fp, model="whisper-1"
        )

    text = transcription.text.strip()

    # [.], [?] 등 단일 특수문자만 들어온 경우 빈 문자열로 처리
    if text in [".", "?", "!", ""]:
        return ""

    return text


def record_audio(
    filename=f"{OUTPUT_DIR}/recorded.wav", duration=5, fs=16000
):
    print("🎤 음성 녹음을 시작합니다... (5초간 말씀하세요)")
    audio_data = sd.rec(int(duration * fs), samplerate=fs, channels=1)
    sd.wait()

    audio_int16 = (audio_data * 32767).astype("int16")
    write(filename, fs, audio_int16)
    print("✅ 녹음 완료:", filename)
    return filename


def ask_gpt(prompt_text):
    print("🤖 GPT 답변 생성 중...")
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "당신은 친절한 20대 여성 AI 비서입니다."},
            {"role": "user", "content": prompt_text},
        ],
    )
    return response.choices[0].message.content.strip()


def main():
    try:
        audio_file = record_audio(duration=5)

        question_text = transcribe_audio(audio_file)
        print(f"\n[사용자 질문]: {question_text}")

        # 음성 입력이 없거나 감지되지 않은 경우 거부
        if not question_text:
            print("⚠️ 음성이 제대로 감지되지 않았습니다. 다시 시도해 주세요.")
            return

        answer = ask_gpt(question_text)
        print(f"\n[GPT 답변]: {answer}\n")

        speak_text(answer)

    except Exception as e:
        print(f"❌ 실행 중 에러가 발생했습니다: {str(e)}")


if __name__ == "__main__":
    main()
````
