# 2. CLIP 실습 과제 모범 답안

## 과제 1. 이미지와 텍스트 임베딩 비교

과제: 이미지 한 장과 텍스트 후보 세 개를 준비하고, CLIP으로 유사도를 비교한다.

예를 들어 이미지에는 강아지가 있고, 후보 문장은 다음과 같다.

* `a photo of a dog`

* `a photo of a cat`

* `a photo of a car`

### 모범 코드

Python

실행됨

```
import torch
from PIL import Image
from transformers import CLIPProcessor, CLIPModel

# 사전 학습된 CLIP 모델을 불러온다.
model_name = "openai/clip-vit-base-patch32"

model = CLIPModel.from_pretrained(model_name)
processor = CLIPProcessor.from_pretrained(model_name)

# 비교할 이미지 파일을 불러온다.
image = Image.open("sample.jpg").convert("RGB")

# 이미지와 비교할 텍스트 후보를 준비한다.
texts = [
    "a photo of a dog",
    "a photo of a cat",
    "a photo of a car",
]

# 이미지와 텍스트를 모델 입력 형식으로 변환한다.
inputs = processor(
    text=texts,
    images=image,
    return_tensors="pt",
    padding=True
)

# 추론 단계에서는 그래디언트를 계산하지 않는다.
with torch.no_grad():
    outputs = model(**inputs)

    # 이미지와 각 텍스트의 유사도 점수를 가져온다.
    scores = outputs.logits_per_image[0]

    # 후보 문장 사이의 상대적인 점수로 변환한다.
    probabilities = scores.softmax(dim=0)

# 점수를 내림차순으로 정렬하여 결과를 확인한다.
ranked = sorted(
    zip(texts, probabilities.tolist()),
    key=lambda item: item[1],
    reverse=True
)

for text, score in ranked:
    print(f"{text}: {score:.4f}")
```

### 모범 결과 해석

강아지 사진에서 `a photo of a dog`가 가장 높은 점수를 얻었다면, 세 후보 중 강아지를 설명하는 텍스트와 이미지가 가장 잘 맞는다고 판단한 것이다.

다만 이 결과만으로 모델이 이미지의 모든 내용을 정확히 이해했다고 결론 내릴 수는 없다. 또한 softmax 점수는 제공된 후보 문장 사이의 상대적인 점수이므로, 출력값이 0.9라고 해서 실제 정답 확률이 90%라는 뜻은 아니다.

## 과제 2. 프롬프트 표현에 따른 분류 결과 비교

과제: 같은 이미지를 서로 다른 텍스트 후보로 분류하고 결과를 비교한다.

예를 들어 다음 두 후보 구성을 비교한다.

| 실험   | 후보 텍스트                                                     |
| ---- | ---------------------------------------------------------- |
| 실험 A | `dog`, `cat`, `car`                                        |
| 실험 B | `a photo of a dog`, `a photo of a cat`, `a photo of a car` |

### 모범 답안

프롬프트 표현이 달라지면 CLIP의 유사도 점수와 순위가 달라질 수 있다. CLIP은 단어 하나만을 기준으로 판단하는 것이 아니라 텍스트 전체의 의미적 표현을 이미지 임베딩과 비교하기 때문이다.

그러나 문장이 길다고 무조건 성능이 좋아지는 것은 아니다. 이미지와 관련성이 높은 표현을 사용하고, 후보 간 문장 형식을 일관되게 유지하는 것이 좋다.

### 추가 실험

다음과 같은 후보 문장을 비교할 수 있다.

* `a dog`

* `a photo of a dog`

* `a photo of a brown dog playing in a park`

이 실험에서는 프롬프트의 구체성이 결과에 어떤 영향을 주는지 관찰한다. 실제 이미지에 없는 특징을 텍스트에 넣으면 오히려 결과가 왜곡될 수 있다.

## 과제 3. MNIST 데이터셋에 CLIP Zero-shot 분류 적용

과제: MNIST 손글씨 숫자 이미지 1,000장을 CLIP으로 분류하고 정확도를 계산한다.

앞서 실험한 결과가 다음과 같다고 가정한다.

* 전체 이미지: 1,000장

* 정답 이미지: 325장

* 오답 이미지: 675장

* 정확도: 32.50%

### 모범 답안

MNIST는 0부터 9까지의 손글씨 숫자를 분류하는 데이터셋이다. 반면 CLIP은 이미지와 자연어 설명 사이의 의미적 대응 관계를 학습한 모델이다. 따라서 MNIST 숫자 이미지와 숫자를 설명하는 텍스트 사이의 표현이 사전 학습된 CLIP의 특징과 잘 맞지 않을 수 있다.

특히 작은 흑백 숫자 이미지의 형태를 구분하는 작업은 자연 이미지의 의미를 텍스트와 연결하는 작업과 다르다. 그러므로 정확도 32.50%는 해당 실험 조건에서 CLIP Zero-shot 분류가 제한적인 성능을 보였다는 의미이다.

개선 방법

1. `the digit 3 handwritten in black on a white background`와 같이 숫자의 시각적 특성을 설명하는 프롬프트를 시험한다.

2. 모든 숫자 후보에 동일한 문장 형식을 적용한다.

3. 원본 이미지의 크기, 채널 수, 배경과 전처리 방식을 점검한다.

4. 숫자별 정확도를 확인하여 어떤 클래스에서 오류가 집중되는지 분석한다.

5. MNIST 전용 CNN 분류기의 결과와 비교한다.

중요한 점은 프롬프트를 수정해도 성능 향상이 보장되지는 않는다는 것이다. MNIST 숫자 분류에서는 해당 데이터셋으로 학습한 CNN이 더 적합한 기준 모델이 될 수 있다.

## 과제 4. CLIP 이미지 검색 구현

과제: 이미지 여러 장과 검색 문장을 입력하고, 텍스트와 가장 유사한 이미지 순위를 출력한다.

### 모범 답안의 핵심 절차

1. 이미지 목록을 준비한다.

2. 각 이미지의 임베딩을 추출한다.

3. 검색 문장의 텍스트 임베딩을 추출한다.

4. 이미지 임베딩과 텍스트 임베딩의 유사도를 계산한다.

5. 유사도가 높은 순서대로 이미지 경로를 정렬한다.

6. 상위 결과를 화면에 표시한다.

예를 들어 검색 문장이 `a person riding a bicycle`이라면 자전거를 타는 사람의 이미지가 상위에 나타나는지 확인한다.

이미지 검색은 분류와 달리 반드시 정해진 클래스 중 하나를 선택할 필요가 없다. 여러 이미지 중 검색 문장과 상대적으로 관련성이 높은 이미지를 찾는 것이 목적이다.

## 과제 5. 이미지 캡셔닝 구현

과제: 이미지 한 장을 입력하여 BLIP이 이미지 설명 문장을 생성하도록 한다.

### 모범 코드

Python

실행됨

```
import torch
from PIL import Image
from transformers import (
    BlipProcessor,
    BlipForConditionalGeneration
)

# 이미지 캡셔닝용 BLIP 사전 학습 모델을 지정한다.
model_name = "Salesforce/blip-image-captioning-base"

# 이미지 전처리와 텍스트 토큰화를 담당하는 프로세서를 불러온다.
processor = BlipProcessor.from_pretrained(model_name)

# 이미지 설명 문장을 생성하는 모델을 불러온다.
model = BlipForConditionalGeneration.from_pretrained(model_name)

# 분석할 이미지를 RGB 형식으로 읽는다.
image = Image.open("sample.jpg").convert("RGB")

# 이미지를 모델이 처리할 수 있는 텐서로 변환한다.
inputs = processor(
    images=image,
    return_tensors="pt"
)

# 평가 시에는 그래디언트 계산을 생략한다.
with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=50
    )

# 생성된 토큰 ID를 사람이 읽을 수 있는 문장으로 변환한다.
caption = processor.decode(
    output_ids[0],
    skip_special_tokens=True
)

print("생성된 이미지 설명:", caption)
```

### 모범 답안

이미지 캡셔닝은 이미지의 시각적 특징을 자연어로 표현하는 작업이다. BLIP은 이미지 정보를 바탕으로 문장을 생성하지만, 생성 문장이 실제 이미지와 완전히 일치한다고 보장할 수는 없다.

예를 들어 이미지에 갈색 강아지가 있는데 모델이 흰색 강아지라고 설명했다면 색상 정보가 잘못 생성된 것이다. 따라서 생성된 문장을 이미지와 비교하여 객체, 색상, 행동, 배경 등이 일치하는지 확인해야 한다.

## 과제 6. BLIP 이미지 캡션 평가

과제: 이미지 5장에 대해 BLIP 캡션을 생성하고 정확성을 평가한다.

### 모범 평가표

| 평가 항목  | 확인할 내용                       |
| ------ | ---------------------------- |
| 객체 정확성 | 설명에 나온 주요 객체가 실제 이미지에 있는가?   |
| 행동 정확성 | 사람이나 동물의 행동을 올바르게 설명했는가?     |
| 속성 정확성 | 색상, 크기, 개수 등의 속성이 적절한가?      |
| 장면 정확성 | 실내, 실외, 배경 등 장면 설명이 맞는가?     |
| 정보 누락  | 이미지의 중요한 내용이 빠지지 않았는가?       |
| 환각 여부  | 이미지에 없는 객체나 사건을 만들어 내지 않았는가? |

### 모범 분석 예시

실제 이미지에 공원에서 공을 가지고 노는 강아지가 있다고 가정한다.

* 생성 결과 A: “A dog is playing with a ball in a park.”

* 생성 결과 B: “A cat is sleeping on a sofa.”

A는 주요 객체와 행동, 배경을 적절하게 설명한다. B는 이미지의 핵심 객체와 장면이 다르므로 부정확한 캡션으로 평가한다.

평가할 때 문장이 자연스러운지만 확인해서는 안 된다. 문장의 자연스러움과 이미지 내용의 정확성은 서로 다른 평가 기준이다.

## 과제 7. BLIP VQA 구현

과제: 이미지와 질문을 입력하고 BLIP이 답변을 생성하도록 한다.

### 모범 코드

Python

실행됨

```
import torch
from PIL import Image
from transformers import (
    BlipProcessor,
    BlipForQuestionAnswering
)

# 시각 질의응답용 BLIP 모델을 지정한다.
model_name = "Salesforce/blip-vqa-base"

# 이미지와 질문을 전처리하는 프로세서를 불러온다.
processor = BlipProcessor.from_pretrained(model_name)

# 이미지 내용을 바탕으로 질문에 답하는 모델을 불러온다.
model = BlipForQuestionAnswering.from_pretrained(model_name)

# 질문할 이미지 파일을 읽는다.
image = Image.open("sample.jpg").convert("RGB")

# 이미지에 대해 질문한다.
question = "What animal is in the picture?"

# 이미지와 질문을 함께 모델 입력으로 변환한다.
inputs = processor(
    images=image,
    text=question,
    return_tensors="pt"
)

# 답변 생성 시에는 그래디언트를 계산하지 않는다.
with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=30
    )

# 토큰 ID를 답변 문장으로 변환한다.
answer = processor.decode(
    output_ids[0],
    skip_special_tokens=True
)

print("질문:", question)
print("답변:", answer)
```

### 모범 답안

VQA는 질문에 포함된 정보를 이미지와 연결하여 답변하는 작업이다. 질문이 “What animal is in the picture?”라면 모델은 이미지에서 동물의 종류를 파악하여 답변을 생성한다.

그러나 모델이 질문의 전제를 무조건 검증하는 것은 아니다. 이미지에 동물이 없더라도 질문의 표현이나 모델의 학습 특성에 따라 잘못된 답변을 생성할 수 있다. 따라서 실제 이미지와 답변이 일치하는지 확인해야 한다.

# 4. CLIP과 BLIP 비교 과제 모범 답안

과제: 같은 이미지에 CLIP과 BLIP을 적용하고 결과를 비교한다.

| 비교 항목 | CLIP                        | BLIP                      |
| ----- | --------------------------- | ------------------------- |
| 주요 목적 | 이미지와 텍스트의 의미적 관련성 비교        | 이미지와 언어의 이해 및 생성          |
| 대표 입력 | 이미지와 비교할 텍스트 후보             | 이미지, 또는 이미지와 질문           |
| 대표 출력 | 텍스트별 유사도 점수                 | 캡션 또는 질문에 대한 답변           |
| 대표 작업 | Zero-shot 분류, 텍스트 기반 이미지 검색 | 이미지 캡셔닝, VQA              |
| 결과 해석 | 후보 중 어느 텍스트가 이미지와 더 관련 있는가? | 이미지에 대해 어떤 문장을 생성했는가?     |
| 주요 한계 | 후보 문장과 도메인에 따라 결과가 달라짐      | 이미지와 일치하지 않는 문장을 생성할 수 있음 |

### 모범 서술형 답안

CLIP과 BLIP은 모두 이미지와 텍스트를 연결하는 멀티모달 모델이지만, 활용 목적에 차이가 있다. CLIP은 이미지와 텍스트를 공통 임베딩 공간에 매핑하여 의미적 유사도를 계산하는 데 적합하다. 따라서 텍스트 후보를 이용한 Zero-shot 분류나 이미지 검색에 활용할 수 있다.

BLIP은 이미지 내용을 자연어로 설명하거나 이미지에 관한 질문에 답하는 데 활용할 수 있다. 그러므로 캡션 생성이나 시각 질의응답에 적합하다.

실제 프로젝트에서는 두 모델을 함께 사용할 수도 있다. BLIP으로 이미지 설명을 생성하고 CLIP으로 설명과 이미지의 의미적 관련성을 비교하는 방식이다. 다만 CLIP의 점수만으로 BLIP 캡션의 사실적 정확성이 입증되는 것은 아니므로, 별도의 정확성 평가가 필요하다.

## 프로젝트 발표 질문과 모범 답안

질문 1. 왜 CLIP과 BLIP을 함께 사용했는가?

모범 답안: BLIP은 이미지 설명을 생성하고 CLIP은 이미지와 텍스트 후보 사이의 의미적 관련성을 비교한다. 두 모델의 기능을 결합하면 이미지에 대한 자연어 설명과 후보별 관련성을 함께 제공할 수 있다.

질문 2. BLIP의 캡션이 틀렸다면 어떻게 개선할 것인가?

모범 답안: 여러 이미지에서 오류 유형을 수집하고 객체, 행동, 색상, 개수 등으로 구분한다. 이미지 전처리와 모델 선택을 점검하고, 필요한 경우 다른 모델과 비교한다. 실제 서비스에서는 생성 결과가 자동으로 사실이라고 간주되지 않도록 검증 절차를 추가한다.

질문 3. CLIP 점수가 높으면 캡션이 정확하다고 판단할 수 있는가?

모범 답안: 그렇지 않다. CLIP 점수는 주어진 이미지와 텍스트의 상대적 관련성을 나타낸다. 잘못된 설명이라도 이미지의 일부 단어와 관련성이 높으면 높은 점수를 얻을 수 있다. 캡션의 정확성은 별도의 기준으로 평가해야 한다.

질문 4. 이 서비스를 실제 업무에 적용할 수 있는 사례는 무엇인가?

모범 답안: 이미지 검색, 상품 이미지 분류 보조, 사진 설명 자동 생성, 이미지 데이터 정리 등에 활용할 수 있다. 다만 의료·안전 등 오류의 영향이 큰 분야에서는 모델 출력만으로 판단하지 않고 사람의 검토와 별도 검증 절차를 적용해야 한다.

# 6. 강사용 최종 정리

수강생이 반드시 이해해야 할 핵심은 다음과 같다.

1. CLIP은 비교하는 모델이다. 이미지와 텍스트를 임베딩 공간에 매핑하여 관련성을 계산한다.

2. BLIP은 생성과 질의응답에 활용하는 모델이다. 이미지 캡션이나 질문에 대한 답변을 생성한다.

3. Zero-shot은 추가적인 작업별 학습 없이 수행하는 방식이다. 그렇다고 모든 데이터에서 높은 정확도를 보장하지는 않는다.

4. 유사도 점수와 정답 확률은 다르다. CLIP의 softmax 점수는 제공된 후보에 대한 상대적인 값이다.

5. 생성 결과는 검증해야 한다. BLIP이 생성한 문장은 자연스러워도 실제 이미지와 다를 수 있다.

6. 모델의 성능은 실제 데이터로 평가해야 한다. 정확도뿐 아니라 클래스별 오류, 캡션의 사실성, 잘못된 답변의 유형을 분석한다.

7. 모델을 결합할 때는 각 모델의 역할을 구분해야 한다. CLIP의 관련성 점수가 BLIP의 생성 문장을 사실로 증명하지는 않는다.

이 내용을 활용하면 강의 중 이론 질문에 대한 답변, 실습 결과 확인, 과제 채점, 프로젝트 발표 평가까지 일관된 기준으로 진행할 수 있다.
