```python
import os
import requests

# 1. 발급받은 API 키 설정
REST_KEY = os.getenv("REST_KEY")

# 2. 최신 v5 엔드포인트 요청
url = "https://api.restcountries.com/countries/v5"
headers = {
    "Authorization": f"Bearer {REST_KEY}"
}
params = {
    "q": "Japan"
}

response = requests.get(url, headers=headers, params=params, timeout=10)

if response.status_code == 200:
    data = response.json()
    japan = data["data"]["objects"][0]
    
    print(f"🏳️ 공식 명칭: {japan['names']['official']}")
    print(f"🏛️ 수도: {japan['capitals'][0]['name']}")
    print(f"👥 인구: {japan['population']:,}명")
else:
    print(f"❌ 데이터 조회 실패: {response.status_code}")
    print(response.text)
```
